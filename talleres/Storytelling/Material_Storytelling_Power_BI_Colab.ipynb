## 1. ¿Qué es storytelling con datos?

El **storytelling con datos** integra tres componentes:

| Componente | Función |
|---|---|
| **Datos** | Proporcionan evidencia y delimitan lo que puede afirmarse. |
| **Visualización** | Hace perceptibles comparaciones, patrones, tendencias y excepciones. |
| **Narrativa** | Organiza la evidencia en una secuencia comprensible para una audiencia y una decisión. |

Una historia con datos no es una colección de gráficos. Es una argumentación sustentada en evidencia: establece un contexto, identifica una tensión o problema, presenta un diagnóstico y conduce a una acción o pregunta de gestión.

Segel y Heer proponen entender la visualización narrativa como una combinación entre estructuras guiadas por el autor e interacción del lector. En Power BI esto puede traducirse en una secuencia guiada mediante páginas, títulos, marcadores y botones, seguida de espacios de exploración mediante filtros, tooltips y drillthrough.

**Fuentes fundamentales**

- Segel, E., & Heer, J. (2010). *Narrative Visualization: Telling Stories with Data*. IEEE Transactions on Visualization and Computer Graphics, 16(6), 1139–1148. https://doi.org/10.1109/TVCG.2010.179
- Knaflic, C. N. (2015). *Storytelling with Data: A Data Visualization Guide for Business Professionals*. Wiley. https://onlinelibrary.wiley.com/doi/book/10.1002/9781119055259
- Schröder, K., et al. (2023). *Telling stories with data: A systematic review*. https://arxiv.org/abs/2312.01164
- Shao, H., et al. (2024). *Data Storytelling in Data Visualisation: Does it Enhance the Efficiency and Effectiveness of Information Retrieval and Insights Comprehension?* https://arxiv.org/abs/2402.12634



## 1.1. Reporte, tablero e historia: no son equivalentes

| Producto | Pregunta principal | Organización | Uso habitual |
|---|---|---|---|
| **Reporte** | ¿Qué información está disponible? | Detalle y consulta | Revisión periódica o auditoría |
| **Tablero** | ¿Cómo se comportan los indicadores? | Monitoreo simultáneo | Seguimiento operativo o estratégico |
| **Historia con datos** | ¿Qué hallazgo importa, por qué importa y qué decisión plantea? | Secuencia deliberada | Comunicación, persuasión informada y decisión |

Un tablero puede contener una historia, pero no la produce automáticamente. Cuando el usuario recibe diez indicadores, seis gráficos y numerosos filtros sin jerarquía, debe descubrir por sí mismo el mensaje. En una narrativa bien diseñada, el autor reduce esa carga cognitiva y orienta la lectura sin ocultar la evidencia.

### Contraejemplo

> “El tablero muestra ventas, costos, productos, países, segmentos y descuentos.”

La frase describe contenido, pero no comunica un hallazgo.

### Ejemplo

> “El segmento Enterprise genera ventas relevantes, pero presenta margen negativo; la prioridad administrativa debe ser revisar su estructura de precio, costo y descuentos.”

La frase establece un hecho, delimita el problema y orienta una decisión.



## 1.2. Marco de construcción: audiencia, pregunta, evidencia, narrativa y acción

### A. Audiencia

Antes de construir el informe, determine:

- ¿Quién tomará la decisión?
- ¿Qué responsabilidad tiene?
- ¿Qué nivel de detalle necesita?
- ¿Qué conocimientos posee sobre los datos?
- ¿Qué interpretación equivocada debe prevenirse?

### B. Pregunta administrativa

Una pregunta útil debe vincular una métrica con una decisión.

**Adecuada**

> ¿Qué segmentos generan o destruyen rentabilidad y en cuáles debe intervenir la gerencia comercial?

**Inadecuada**

> ¿Cuáles son las ventas?

La segunda pregunta es descriptiva y demasiado amplia. Puede ser parte del análisis, pero no define una decisión.

### C. Evidencia

La evidencia debe incluir:

- magnitud;
- comparación;
- contexto temporal;
- desagregación;
- excepciones;
- limitaciones.

### D. Narrativa

Una estructura aplicable a informes ejecutivos es:

1. **Contexto:** cuál es el desempeño general.
2. **Tensión:** dónde aparece una desviación o riesgo.
3. **Diagnóstico:** qué segmentos, productos o condiciones concentran el problema.
4. **Explicación prudente:** qué variables están asociadas con el resultado.
5. **Acción:** qué debe revisarse, por quién y mediante qué indicador.

### E. Acción

La recomendación debe ser específica y medible. No debe exceder lo que los datos permiten afirmar.



## 1.3. Principios de percepción y diseño visual

La selección de un gráfico debe responder a una **tarea analítica**, no a una preferencia estética.

| Tarea | Visualización generalmente apropiada | Uso incorrecto frecuente |
|---|---|---|
| Comparar categorías | Barras ordenadas | Circular con muchas categorías |
| Analizar evolución temporal | Línea | Barras 3D o categorías sin orden temporal |
| Relacionar dos variables | Dispersión | Dos gráficos separados sin escala comparable |
| Mostrar composición simple | Barras apiladas o 100 % apiladas | Circular con porciones similares |
| Identificar detalle | Tabla o matriz | Incluir todos los registros en la página ejecutiva |
| Resaltar una excepción | Anotación y contraste selectivo | Colorear cada elemento de forma diferente |

Cleveland y McGill demostraron que las personas estiman con mayor precisión posiciones sobre una escala común que ángulos, áreas o volúmenes. Por ello, las barras y puntos suelen facilitar comparaciones más precisas que gráficos circulares, burbujas o elementos tridimensionales.

**Fuentes**

- Cleveland, W. S., & McGill, R. (1984). *Graphical Perception: Theory, Experimentation, and Application to the Development of Graphical Methods*. Journal of the American Statistical Association, 79(387), 531–554. https://doi.org/10.1080/01621459.1984.10478080
- Munzner, T. (2014). *Visualization Analysis and Design*. CRC Press. https://www.cs.ubc.ca/~tmm/vadbook/
- Tufte, E. R. (2001). *The Visual Display of Quantitative Information* (2nd ed.). Graphics Press. https://www.edwardtufte.com/books/



## 1.4. Integridad analítica y ética

Una narrativa no debe convertir una asociación en una causa.

| Nivel de afirmación | Ejemplo |
|---|---|
| **Descriptiva** | Enterprise presenta una utilidad acumulada negativa. |
| **Comparativa** | Su margen es inferior al de los demás segmentos. |
| **Relacional** | Dentro de Enterprise, las bandas de mayor descuento coinciden con márgenes más bajos. |
| **Causal** | Los descuentos causaron las pérdidas de Enterprise. |

El último enunciado exige controlar otras variables: costos, productos, condiciones comerciales, periodos y composición de clientes. Con un análisis descriptivo no es posible demostrar causalidad.

### Prácticas incorrectas

- Seleccionar solo periodos que favorecen una conclusión.
- Ocultar categorías con resultados adversos.
- Truncar ejes para exagerar diferencias.
- usar porcentajes sin indicar el denominador.
- Comparar periodos con diferente cobertura.
- presentar datos simulados como resultados reales.
- afirmar causalidad sin diseño experimental o evidencia suficiente.
- utilizar una narración automática sin validar sus cálculos y supuestos.



## 2. Caso y conjunto de datos

Se utilizará el conjunto oficial **Financial Sample** de Microsoft. Contiene ventas, descuentos, costos y beneficios por producto, segmento y país o región.

**Fuente oficial:**  
https://learn.microsoft.com/es-es/power-bi/create-reports/sample-financial-download

**Descarga directa:**  
https://download.microsoft.com/download/1/4/E/14EDED28-6C58-4055-A65C-23B4DA81C4DE/Financial%20Sample.xlsx

### Pregunta administrativa del caso

> ¿Qué segmento requiere intervención prioritaria para proteger la rentabilidad y qué evidencia debe presentar la gerencia comercial?

### Hipótesis preliminares que deberán evaluarse

1. Los segmentos con mayores ventas son necesariamente los más rentables.
2. Los descuentos explican por sí solos el deterioro de la rentabilidad.
3. Los años 2013 y 2014 pueden compararse directamente.
4. Un producto con ventas elevadas siempre debe recibir mayor inversión.

Estas proposiciones no se asumirán como verdaderas. Se contrastarán con los datos.


````python

# Descargar el conjunto oficial de Microsoft.
# Si la conexión no está disponible, se reconstruye una copia de respaldo
# incluida en el cuaderno a partir del mismo conjunto oficial.

from pathlib import Path
import requests
import gzip
import base64

URL = (
    "https://download.microsoft.com/download/1/4/E/"
    "14EDED28-6C58-4055-A65C-23B4DA81C4DE/Financial%20Sample.xlsx"
)
archivo_excel = Path("Financial_Sample.xlsx")
archivo_respaldo = Path("Financial_Sample_fallback.csv")

RESPALDO_B64 = """"""

try:
    respuesta = requests.get(URL, timeout=30)
    respuesta.raise_for_status()
    archivo_excel.write_bytes(respuesta.content)
    archivo_datos = archivo_excel
    print(f"Archivo descargado: {archivo_datos.resolve()}")
except Exception as error:
    contenido = gzip.decompress(base64.b64decode(RESPALDO_B64))
    archivo_respaldo.write_bytes(contenido)
    archivo_datos = archivo_respaldo
    print("No fue posible descargar el archivo; se utilizó la copia de respaldo incluida.")
    print(f"Archivo reconstruido: {archivo_datos.resolve()}")
    print("Detalle técnico de la descarga:", error)
````

````python

import pandas as pd
import numpy as np
from IPython.display import display

if archivo_datos.suffix.lower() == ".xlsx":
    datos = pd.read_excel(archivo_datos)
else:
    datos = pd.read_csv(archivo_datos)

datos.columns = [col.strip().replace(" ", "_") for col in datos.columns]
datos["Date"] = pd.to_datetime(datos["Date"])

print("Dimensiones:", datos.shape)
display(datos.head())
````


## 2.1. Diccionario básico

| Campo | Significado |
|---|---|
| `Segment` | Segmento comercial |
| `Country` | País o región |
| `Product` | Producto |
| `Discount_Band` | Banda de descuento |
| `Units_Sold` | Unidades vendidas |
| `Manufacturing_Price` | Precio de fabricación registrado |
| `Sale_Price` | Precio de venta de referencia |
| `Gross_Sales` | Ventas antes de descuentos |
| `Discounts` | Valor de descuentos |
| `Sales` | Ventas netas |
| `COGS` | Costo de los bienes vendidos |
| `Profit` | Utilidad |
| `Date` | Fecha mensual |
| `Year` | Año |

### Relaciones contables que deben verificarse

\[
Ventas\ netas \approx Ventas\ brutas - Descuentos
\]

\[
Utilidad \approx Ventas\ netas - Costo\ de\ ventas
\]

La aproximación reconoce posibles diferencias mínimas por redondeo.


````python

# Crear métricas derivadas útiles para la narrativa.

datos["Margin"] = np.where(
    datos["Sales"] != 0,
    datos["Profit"] / datos["Sales"],
    np.nan
)

datos["Discount_Rate"] = np.where(
    datos["Gross_Sales"] != 0,
    datos["Discounts"] / datos["Gross_Sales"],
    np.nan
)

datos["Year_Month"] = datos["Date"].dt.to_period("M").astype(str)

# Verificaciones de consistencia.
datos["Sales_Check"] = datos["Gross_Sales"] - datos["Discounts"]
datos["Profit_Check"] = datos["Sales"] - datos["COGS"]

revision = pd.DataFrame({
    "Indicador": [
        "Diferencia máxima: Sales vs Gross Sales - Discounts",
        "Diferencia máxima: Profit vs Sales - COGS",
        "Valores faltantes totales",
        "Registros duplicados completos"
    ],
    "Resultado": [
        (datos["Sales"] - datos["Sales_Check"]).abs().max(),
        (datos["Profit"] - datos["Profit_Check"]).abs().max(),
        int(datos.isna().sum().sum()),
        int(datos.duplicated().sum())
    ]
})

display(revision)
````


## 2.2. Cobertura temporal: un control obligatorio

Antes de comparar años, se debe comprobar si cada periodo contiene el mismo número de meses. Una variación anual calculada entre un año completo y un año parcial sería matemáticamente correcta, pero administrativamente engañosa.


````python

cobertura = (
    datos.groupby("Year")
    .agg(
        Fecha_inicial=("Date", "min"),
        Fecha_final=("Date", "max"),
        Meses_disponibles=("Date", "nunique"),
        Registros=("Date", "size")
    )
    .reset_index()
)

display(cobertura)
````


### Interpretación correcta

El conjunto contiene únicamente septiembre–diciembre de 2013 y los doce meses de 2014. Por tanto, **no es válido presentar el crecimiento 2014 frente a 2013 como una comparación anual homogénea**.

### Contraejemplo

> “Las ventas crecieron 249 % entre 2013 y 2014.”

Aunque el cálculo numérico puede obtenerse, la conclusión es inadecuada porque compara cuatro meses con doce meses.

### Alternativas rigurosas

- Limitar la comparación interanual a septiembre–diciembre.
- Analizar 2014 por separado.
- Evitar la comparación temporal si no es necesaria para la pregunta de negocio.

En este caso se analizará la rentabilidad acumulada por segmento, sin interpretar 2013 y 2014 como años comparables.


````python

# Resumen general del conjunto.

resumen_general = pd.DataFrame({
    "Indicador": [
        "Ventas netas",
        "Utilidad",
        "Margen",
        "Unidades",
        "Descuentos"
    ],
    "Valor": [
        datos["Sales"].sum(),
        datos["Profit"].sum(),
        datos["Profit"].sum() / datos["Sales"].sum(),
        datos["Units_Sold"].sum(),
        datos["Discounts"].sum()
    ]
})

display(resumen_general)
````


## 3. Exploración orientada a la pregunta administrativa

No se construirán gráficos al azar. La secuencia analítica será:

1. Determinar qué segmentos generan utilidad.
2. Identificar si alguno destruye valor pese a sus ventas.
3. Examinar la economía por unidad.
4. Evaluar la relación entre descuentos y margen dentro del segmento crítico.
5. Identificar productos que concentran la pérdida.
6. Formular una recomendación prudente.


````python

segmentos = (
    datos.groupby("Segment")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Units=("Units_Sold", "sum"),
        Discounts=("Discounts", "sum"),
        Gross_Sales=("Gross_Sales", "sum"),
        COGS=("COGS", "sum")
    )
    .reset_index()
)

segmentos["Margin"] = segmentos["Profit"] / segmentos["Sales"]
segmentos["Discount_Rate"] = segmentos["Discounts"] / segmentos["Gross_Sales"]
segmentos = segmentos.sort_values("Profit", ascending=True)

display(
    segmentos.style.format({
        "Sales": "${:,.0f}",
        "Profit": "${:,.0f}",
        "Discounts": "${:,.0f}",
        "Gross_Sales": "${:,.0f}",
        "COGS": "${:,.0f}",
        "Margin": "{:.1%}",
        "Discount_Rate": "{:.1%}"
    })
)
````

````python

# Visual principal: utilidad por segmento.
# Una barra facilita comparar magnitudes sobre una escala común.

import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 5))
ax.barh(segmentos["Segment"], segmentos["Profit"])
ax.axvline(0, linewidth=1)
ax.set_title("Enterprise es el único segmento con utilidad acumulada negativa")
ax.set_xlabel("Utilidad")
ax.set_ylabel("Segmento")

for i, valor in enumerate(segmentos["Profit"]):
    ax.text(
        valor,
        i,
        f" {valor:,.0f}",
        va="center",
        ha="left" if valor >= 0 else "right"
    )

plt.tight_layout()
plt.show()
````


### Lectura correcta

- Enterprise registra ventas cercanas a 19,6 millones, pero una utilidad aproximada de **−614,5 mil**.
- Su margen es cercano a **−3,1 %**.
- El volumen de ventas no garantiza creación de valor.

### Título narrativo adecuado

> **Enterprise vende 19,6 millones, pero destruye aproximadamente 614 mil de utilidad.**

### Título inadecuado

> **Utilidad por segmento**

El segundo título solo nombra las variables. No comunica el hallazgo.

### Contraejemplo administrativo

> “Enterprise es uno de los segmentos con mayores ventas; por tanto, debe recibir más inversión comercial.”

La recomendación ignora la rentabilidad. Antes de invertir, debe comprenderse la economía unitaria y la composición de la pérdida.


````python

# Economía por unidad: permite distinguir volumen de rentabilidad.

segmentos["Net_Sales_per_Unit"] = segmentos["Sales"] / segmentos["Units"]
segmentos["COGS_per_Unit"] = segmentos["COGS"] / segmentos["Units"]
segmentos["Profit_per_Unit"] = segmentos["Profit"] / segmentos["Units"]
segmentos["Discount_per_Unit"] = segmentos["Discounts"] / segmentos["Units"]

columnas = [
    "Segment",
    "Net_Sales_per_Unit",
    "COGS_per_Unit",
    "Profit_per_Unit",
    "Margin"
]

display(
    segmentos[columnas]
    .sort_values("Profit_per_Unit")
    .style.format({
        "Net_Sales_per_Unit": "${:,.2f}",
        "COGS_per_Unit": "${:,.2f}",
        "Profit_per_Unit": "${:,.2f}",
        "Margin": "{:.1%}"
    })
)
````


### Hallazgo de economía unitaria

En Enterprise:

- ventas netas por unidad: aproximadamente **116,35**;
- costo por unidad: **120,00**;
- utilidad por unidad: aproximadamente **−3,65**.

Esto indica que el problema no se explica únicamente por el volumen. En promedio, la unidad vendida no cubre su costo de ventas.

### Formulación rigurosa

> La estructura observada de precio neto y costo produce una utilidad unitaria negativa en Enterprise.

### Formulación no sustentada

> Los vendedores de Enterprise están gestionando mal a los clientes.

El dataset no contiene información sobre desempeño individual, negociaciones, comisiones o calidad de gestión. Esa afirmación excede la evidencia disponible.


````python

# Examinar bandas de descuento dentro de Enterprise.

enterprise = datos.loc[datos["Segment"] == "Enterprise"].copy()

enterprise_descuento = (
    enterprise.groupby("Discount_Band")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Units=("Units_Sold", "sum"),
        Discounts=("Discounts", "sum"),
        Gross_Sales=("Gross_Sales", "sum"),
        COGS=("COGS", "sum")
    )
    .reset_index()
)

enterprise_descuento["Margin"] = (
    enterprise_descuento["Profit"] / enterprise_descuento["Sales"]
)
enterprise_descuento["Discount_Rate"] = (
    enterprise_descuento["Discounts"] / enterprise_descuento["Gross_Sales"]
)

orden = pd.CategoricalDtype(["Low", "Medium", "High"], ordered=True)
enterprise_descuento["Discount_Band"] = (
    enterprise_descuento["Discount_Band"].astype(orden)
)
enterprise_descuento = enterprise_descuento.sort_values("Discount_Band")

display(
    enterprise_descuento.style.format({
        "Sales": "${:,.0f}",
        "Profit": "${:,.0f}",
        "Discounts": "${:,.0f}",
        "Gross_Sales": "${:,.0f}",
        "COGS": "${:,.0f}",
        "Margin": "{:.1%}",
        "Discount_Rate": "{:.1%}"
    })
)
````

````python

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(
    enterprise_descuento["Discount_Band"].astype(str),
    enterprise_descuento["Margin"]
)
ax.axhline(0, linewidth=1)
ax.set_title("En Enterprise, las bandas de mayor descuento coinciden con menor margen")
ax.set_xlabel("Banda de descuento")
ax.set_ylabel("Margen")

for i, valor in enumerate(enterprise_descuento["Margin"]):
    ax.text(i, valor, f"{valor:.1%}", ha="center", va="bottom" if valor >= 0 else "top")

plt.tight_layout()
plt.show()
````


### Interpretación prudente

- Banda baja: margen ligeramente positivo.
- Banda media: margen negativo.
- Banda alta: margen aún más negativo.
- La evidencia muestra una **asociación** entre mayor descuento y menor margen dentro de Enterprise.
- Sin embargo, incluso la banda baja se encuentra cerca del punto de equilibrio. Reducir descuentos podría ayudar, pero probablemente no sea suficiente si no se revisan precio y costo.

### Contraejemplo causal

> “Los descuentos son la causa única de las pérdidas.”

No es correcto. La información muestra que el costo por unidad también es superior a las ventas netas por unidad. Además, la composición por producto puede influir.


````python

# Productos que concentran las pérdidas dentro de Enterprise.

enterprise_productos = (
    enterprise.groupby("Product")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Units=("Units_Sold", "sum"),
        Discounts=("Discounts", "sum"),
        COGS=("COGS", "sum")
    )
    .reset_index()
)

enterprise_productos["Margin"] = (
    enterprise_productos["Profit"] / enterprise_productos["Sales"]
)
enterprise_productos = enterprise_productos.sort_values("Profit")

display(
    enterprise_productos.style.format({
        "Sales": "${:,.0f}",
        "Profit": "${:,.0f}",
        "Discounts": "${:,.0f}",
        "COGS": "${:,.0f}",
        "Margin": "{:.1%}"
    })
)
````

````python

fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(enterprise_productos["Product"], enterprise_productos["Profit"])
ax.axvline(0, linewidth=1)
ax.set_title("Carretera concentra la mayor pérdida dentro de Enterprise")
ax.set_xlabel("Utilidad")
ax.set_ylabel("Producto")

for i, valor in enumerate(enterprise_productos["Profit"]):
    ax.text(valor, i, f" {valor:,.0f}", va="center", ha="right")

plt.tight_layout()
plt.show()
````


## 4. Síntesis narrativa basada en evidencia

### Contexto

El conjunto registra rentabilidad positiva en términos agregados.

### Tensión

Enterprise es el único segmento con utilidad acumulada negativa, pese a registrar cerca de 19,6 millones en ventas.

### Diagnóstico

- Margen de Enterprise: aproximadamente −3,1 %.
- Utilidad por unidad: aproximadamente −3,65.
- Ventas netas por unidad: aproximadamente 116,35.
- Costo por unidad: 120,00.
- Las bandas media y alta de descuento presentan márgenes negativos.
- Carretera concentra la mayor pérdida dentro del segmento.

### Recomendación administrativa

> Revisar la estructura de precio, costo y autorización de descuentos del segmento Enterprise, priorizando las operaciones de banda alta y el producto Carretera. El seguimiento debe realizarse mediante margen, utilidad por unidad y proporción de ventas bajo descuentos altos.

### Limitación

El análisis es descriptivo. No identifica causalidad ni contiene información sobre contratos, clientes, vendedores, logística o calidad. Estas variables deberían incorporarse antes de adoptar una política definitiva.

```DAX
SUM(financials[Gross Sales])

Descuentos =
SUM(financials[Discounts])

Costo de Ventas =
SUM(financials[COGS])

Utilidad =
SUM(financials[Profit])

Unidades =
SUM(financials[Units Sold])

Margen % =
DIVIDE([Utilidad], [Ventas])

Tasa de Descuento % =
DIVIDE([Descuentos], [Ventas Brutas])

Ventas por Unidad =
DIVIDE([Ventas], [Unidades])

Costo por Unidad =
DIVIDE([Costo de Ventas], [Unidades])

Utilidad por Unidad =
DIVIDE([Utilidad], [Unidades])
```

### Nota de rigor

No calcule el margen como promedio de la columna de márgenes por fila. El margen agregado debe calcularse como:

\[
\frac{\sum Utilidad}{\sum Ventas}
\]

El promedio simple de porcentajes puede asignar el mismo peso a observaciones de tamaños diferentes.

## 5.4. Elementos de Power BI con función narrativa

### Marcadores

Los marcadores capturan el estado de filtros, segmentadores, orden y visibilidad. Puede crear tres estados:

1. Panorama general.
2. Enterprise.
3. Enterprise con banda alta.

Estos estados permiten conducir la exposición sin duplicar páginas.

Documentación oficial:  
https://learn.microsoft.com/en-us/power-bi/create-reports/desktop-bookmarks

### Navegador de marcadores o páginas

Utilícelo para hacer visible la secuencia. Los nombres deben comunicar una etapa, por ejemplo: **Panorama → Diagnóstico → Acción**.

Documentación oficial:  
https://learn.microsoft.com/en-us/power-bi/create-reports/button-navigators

### Tooltip de página

Úselo para proporcionar detalle contextual al pasar el cursor. No debe contener información esencial que solo pueda descubrirse accidentalmente.

Documentación oficial:  
https://learn.microsoft.com/en-us/power-bi/create-reports/desktop-tooltips

### Drillthrough

Úselo cuando el lector necesite pasar de una visión resumida al detalle filtrado de un segmento o producto.

Documentación oficial:  
https://learn.microsoft.com/en-us/power-bi/create-reports/desktop-drillthrough



## 5.5. Narración inteligente: apoyo, no sustitución del juicio

Power BI puede generar resúmenes automáticos mediante **Smart narrative**. Esta función puede ayudar a redactar un primer texto, pero el estudiante debe verificar:

- cifras;
- contexto de filtros;
- cobertura temporal;
- denominadores;
- excepciones;
- causalidad;
- coherencia con la pregunta administrativa.

### Uso correcto

> Generar una narración preliminar y editarla para incluir el contexto de Enterprise, la cobertura parcial de 2013 y la diferencia entre asociación y causalidad.

### Uso incorrecto

> Insertar el texto automático como conclusión final sin contrastarlo con las medidas.

Documentación oficial:  
https://learn.microsoft.com/en-us/power-bi/visuals/power-bi-visualization-smart-narrative



## 5.6. Accesibilidad y legibilidad

Un informe riguroso también debe ser utilizable por diferentes personas.

Verifique:

- contraste suficiente;
- tamaño de texto legible;
- títulos y etiquetas explícitas;
- texto alternativo en visuales;
- orden de tabulación lógico;
- no depender exclusivamente del color;
- ausencia de fondos decorativos que reduzcan legibilidad;
- unidades de medida visibles;
- formato consistente de moneda y porcentaje.

Documentación oficial:  
https://learn.microsoft.com/en-us/power-bi/create-reports/desktop-accessibility-creating-reports
